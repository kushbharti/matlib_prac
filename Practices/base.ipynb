```python
import matplotlib.pyplot as plt
import pandas as pd 
import seaborn as sns
```

```python
data = pd.read_csv('final_vg.csv')
data.head()
```

```python
data['Genre'].unique()
```

```python
category_count = data['Genre'].value_counts()
```

```python
x_val = category_count.index
y_val = category_count.values

plt.figure(figsize=(12,8))
plt.bar(x_val, y_val, width=0.7, label='Genre Count')
plt.xticks(rotation=90, fontsize=10)
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Top Genre Analysis')
plt.legend()
plt.show()
```

```python
sales_data = data[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']]
region_sales = sales_data.T.sum(axis='columns')


plt.figure(figsize=(10,6))
plt.pie(region_sales, labels=region_sales.index, startangle=90, autopct='%1.1f%%', explode=(0.07,0,0,0))
plt.title('Total Sales across various regions')
plt.show()
```

```python
plt.figure(figsize=(11,7))
plt.hist(data['Year'], bins=15, edgecolor='black')
plt.show()
```
